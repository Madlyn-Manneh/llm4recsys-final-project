# Session 2 RQ2 reproduction
One fresh Linux Python 3.11.5 process per archived source notebook; one idle H100 for inference. Source logic and prompts are unchanged. This post-experiment launcher is not a second measured experiment.

```python
from pathlib import Path
import os, sys, json, hashlib, subprocess
ROOT=Path.cwd()
OWN=ROOT/'llm4rec_s2_rq2'
assert not (OWN/'recovered_v1').exists(), 'Preserve old results; use a fresh copy for a new run.'
def run_source(name, stop_before=None):
    path=ROOT/'notebooks'/name
    nb=json.loads(path.read_text())
    code=''.join(''.join(c['source']) for c in nb['cells'] if c['cell_type']=='code')
    if stop_before:
        assert code.count(stop_before)==1
        code=code.split(stop_before)[0]
    env=dict(os.environ,CART_LAUNCH_SOURCE_SHA256=hashlib.sha256(code.encode()).hexdigest())
    print('Running',name,flush=True)
    subprocess.run([sys.executable,'-c',code],cwd=ROOT,env=env,check=True)
run_source('setup_inputs_v1.ipynb')
run_source('llm4rec_s2_rq2_prepare_v2.ipynb')
run_source('llm4rec_s2_rq2_seeded_baseline_v1.ipynb')
# Preserve the executed independent verification code, omitting only its historical blocked-network status report.
run_source('llm4rec_s2_rq2_verify_preparation_v2.ipynb',stop_before='network=json.loads')
run_source('llm4rec_s2_rq2_offline_inputcheck_v1.ipynb')
run_source('llm4rec_s2_rq2_inference_v4.ipynb')
run_source('llm4rec_s2_rq2_evaluate_v2.ipynb')
print('Results:',OWN/'recovered_v1/evaluation/results-summary.json')
```
